```python
def google_places_search(query, api_key, max_pages=3):
    url = "https://places.googleapis.com/v1/places:searchText"

    field_mask = ",".join([
        "places.id",
        "places.displayName",
        "places.formattedAddress",
        "places.location",
        "places.rating",
        "places.userRatingCount",
        "places.googleMapsUri",
        "places.websiteUri",
        "places.businessStatus",
        "places.primaryType",
        "places.nationalPhoneNumber", # Added to retrieve phone number
        "nextPageToken"
    ])

    headers = {
        "Content-Type": "application/json",
        "X-Goog-Api-Key": api_key,
        "X-Goog-FieldMask": field_mask
    }

    results = []
    page_token = None

    for page in range(max_pages):

        body = {
            "textQuery": query,
            "languageCode": "en",
            "regionCode": "IN",
            "pageSize": 20
        }

        if page_token:
            body["pageToken"] = page_token

        response = requests.post(
            url,
            headers=headers,
            json=body,
            timeout=30
        )

        print(
            f"Page {page + 1}: HTTP {response.status_code}"
        )

        if response.status_code != 200:
            print(response.text)
            break

        data = response.json()

        places = data.get("places", [])

        results.extend(places)

        print(
            f"Found {len(places)} places"
        )

        page_token = data.get("nextPageToken")

        if not page_token:
            break

        # Allow the next page token to become usable.
        time.sleep(2)

    return results
```

```python
# ---------- 3. CREATE LEADS from all_places ----------
rows = []

for p in all_places:

    rating = p.get("rating", 0)
    reviews = p.get("userRatingCount", 0)
    website = p.get("websiteUri", "")
    category = p.get("Category", "") # Retrieve category
    city = p.get("City", "") # Retrieve city
    phone_number = p.get("nationalPhoneNumber", "") # Retrieve phone number

    score = 0

    # Reviews
    if reviews >= 500:
        score += 20
    elif reviews >= 200:
        score += 15
    elif reviews >= 50:
        score += 10
    elif reviews >= 10:
        score += 5

    # Rating
    if rating >= 4.7:
        score += 15
    elif rating >= 4.4:
        score += 10
    elif rating >= 4.0:
        score += 5

    # Website opportunity
    if not website:
        score += 30
    else:
        score += 5

    # Operational
    if p.get("businessStatus") == "OPERATIONAL":
        score += 10

    score = min(score, 100)

    if score >= 80:
        priority = "HOT"
    elif score >= 60:
        priority = "HIGH"
    elif score >= 40:
        priority = "MEDIUM"
    else:
        priority = "LOW"

    rows.append({
        "Category": category,
        "City": city,
        "Place ID": p.get("id", ""),
        "Business Name": p.get("displayName", {}).get("text", ""),
        "Address": p.get("formattedAddress", ""),
        "Google Maps": p.get("googleMapsUri", ""),
        "Rating": rating,
        "Reviews": reviews,
        "Website": website,
        "Website Missing": "YES" if not website else "NO",
        "Phone": phone_number, # Assign retrieved phone number
        "Email": '', # Requires web scraping
        "WhatsApp": '', # Requires web scraping
        "Instagram": '', # Requires web scraping
        "Facebook": '', # Requires web scraping (added for completeness, though not in user's final list)
        "LinkedIn": '', # Requires web scraping (added for completeness)
        "Owner/Decision Maker": '', # Requires web scraping
        "Services": '', # Requires web scraping
        "Pricing/Packages": '', # Requires web scraping
        "Booking Available": False, # Requires web scraping
        "Website Quality": None, # Requires custom analysis
        "Lead Score": score, # Renamed from Code Bank Score
        "Priority": priority,
        "Contact Status": '', # CRM Field
        "Follow-up": '', # CRM Field
        "Deal Value": None # CRM Field
    })
```

```python
# ---------- 4. DATAFRAME ----------
df_combined = pd.DataFrame(rows)

# Remove duplicates based on 'Place ID'
if not df_combined.empty:
    df_combined = df_combined.drop_duplicates(subset=["Place ID"])
    df_combined = df_combined.sort_values("Lead Score", ascending=False)

# ---------- 5. DISPLAY ----------
print("\n================================")
print(" COMBINED MULTI-CATEGORY LEADS")
print("================================")
print(f"Total unique leads found: {len(df_combined)}")

display(df_combined[[
    "Business Name",
    "Category",
    "City",
    "Google Maps",
    "Website",
    "Phone",
    "Email",
    "WhatsApp",
    "Instagram",
    "Owner/Decision Maker",
    "Rating",
    "Reviews",
    "Website Missing",
    "Website Quality",
    "Booking Available",
    "Lead Score",
    "Priority",
    "Contact Status",
    "Follow-up",
    "Deal Value"
]].head(10)) # Display top 10 leads with all requested columns
```

```python
# ============================================================
# CODE BANK - ONE CELL LEAD FINDER TEST
# Google Colab + Google Places API (New)
# ============================================================

!pip -q install requests pandas

import requests
import pandas as pd
import time
from getpass import getpass
from IPython.display import display
from google.colab import files

# ---------- 1. SETTINGS ----------
API_KEY = getpass("Paste your Google Places API Key: ")

CATEGORY = "Dental Clinic"
CITY = "Pune"

QUERY = f"{CATEGORY} in {CITY}, India"

# ---------- 2. GOOGLE PLACES SEARCH ----------
url = "https://places.googleapis.com/v1/places:searchText"

headers = {
    "Content-Type": "application/json",
    "X-Goog-Api-Key": API_KEY,
    "X-Goog-FieldMask": ",".join([
        "places.id",
        "places.displayName",
        "places.formattedAddress",
        "places.location",
        "places.rating",
        "places.userRatingCount",
        "places.googleMapsUri",
        "places.websiteUri",
        "places.businessStatus",
        "places.primaryType"
    ])
}

body = {
    "textQuery": QUERY,
    "languageCode": "en",
    "regionCode": "IN",
    "pageSize": 20
}

response = requests.post(
    url,
    headers=headers,
    json=body,
    timeout=30
)

print("HTTP STATUS:", response.status_code)

if response.status_code != 200:
    print(response.text)
    raise Exception("Google Places API request failed.")

data = response.json()
places = data.get("places", [])

print("Businesses found:", len(places))

# ---------- 3. CREATE LEADS ----------
rows = []

for p in places:

    rating = p.get("rating", 0)
    reviews = p.get("userRatingCount", 0)
    website = p.get("websiteUri", "")

    score = 0

    # Reviews
    if reviews >= 500:
        score += 20
    elif reviews >= 200:
        score += 15
    elif reviews >= 50:
        score += 10
    elif reviews >= 10:
        score += 5

    # Rating
    if rating >= 4.7:
        score += 15
    elif rating >= 4.4:
        score += 10
    elif rating >= 4.0:
        score += 5

    # Website opportunity
    if not website:
        score += 30
    else:
        score += 5

    # Operational
    if p.get("businessStatus") == "OPERATIONAL":
        score += 10

    score = min(score, 100)

    if score >= 80:
        priority = "HOT"
    elif score >= 60:
        priority = "HIGH"
    elif score >= 40:
        priority = "MEDIUM"
    else:
        priority = "LOW"

    rows.append({
        "Category": CATEGORY,
        "City": CITY,
        "Place ID": p.get("id", ""),
        "Business Name": p.get(
            "displayName", {}
        ).get("text", ""),
        "Address": p.get(
            "formattedAddress", ""
        ),
        "Rating": rating,
        "Reviews": reviews,
        "Website": website,
        "Website Missing": "YES" if not website else "NO",
        "Google Maps": p.get(
            "googleMapsUri", ""
        ),
        "Status": p.get(
            "businessStatus", ""
        ),
        "Type": p.get(
            "primaryType", ""
        ),
        "Code Bank Score": score,
        "Priority": priority
    })

# ---------- 4. DATAFRAME ----------
df = pd.DataFrame(rows)

# Remove duplicates
if not df.empty:
    df = df.drop_duplicates(
        subset=["Place ID"]
    )

    df = df.sort_values(
        "Code Bank Score",
        ascending=False
    )

# ---------- 5. DISPLAY ----------
print("\n================================")
print(" CODE BANK LEAD TEST")
print("================================")
print("Query:", QUERY)
print("Unique leads:", len(df))

display(df)

# ---------- 6. HOT LEADS ----------
print("\n🔥 HIGH-VALUE CODE BANK LEADS")

hot = df[
    df["Code Bank Score"] >= 60
]

display(
    hot[
        [
            "Business Name",
            "Rating",
            "Reviews",
            "Website Missing",
            "Code Bank Score",
            "Priority",
            "Google Maps"
        ]
    ]
)

# ---------- 7. SAVE CSV ----------
filename = (
    "codebank_"
    + CATEGORY.lower().replace(" ", "_")
    + "_"
    + CITY.lower().replace(" ", "_")
    + ".csv"
)

df.to_csv(
    filename,
    index=False,
    encoding="utf-8-sig"
)

print("\nCSV CREATED:", filename)

# ---------- 8. DOWNLOAD ----------
if len(df) > 0:
    files.download(filename)

print("\n✅ TEST COMPLETE")
```
